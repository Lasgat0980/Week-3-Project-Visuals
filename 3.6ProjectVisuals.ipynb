```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from IPython.display import display, Markdown
from datetime import datetime


todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "Lasgat0980"

file_path = r"C:\Users\ldick\Downloads\StudentsPerformance.xlsx"

df = pd.read_excel(file_path)

df.head()
```

```python
#Get current date and time
current_datetime = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

#Create Histogram of Math Scores
plt.figure(figsize=(10, 6))
plt.hist(df['math score'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.title(f'Histogram of Math Scores - StudentID: {student_id} - {current_datetime}')
plt.grid(True)
plt.show()
```

```python
#Calculate average math score by race/ethnicity
average_math_by_race = df.groupby('race/ethnicity')['math score'].mean()

#Display averages
display(average_math_by_race)

#Create Bar Chart
plt.figure(figsize=(10, 6))
plt.bar(average_math_by_race.index, average_math_by_race.values, color='green')
plt.xlabel('Race/Ethnicity')
plt.ylabel('Average Math Score')
plt.title(f'Average Math Score by Race/Ethnicity - StudentID: {student_id} - {current_datetime}')
plt.grid(True)
plt.show()
```

```python
#Create Scatter Plot of Math Score vs Writing Score
x = df['math score']
y = df['writing score']

z = np.polyfit(x, y, 1)
p = np.poly1d(z)

plt.figure(figsize=(12, 8))
plt.scatter(x, y, label='Data Points', color='blue')
plt.plot(x, p(x), label=f'Trendline: y = {z[0]:.2f}x + {z[1]:.2f}', color='red')
plt.xlabel('Math Score')
plt.ylabel('Writing Score')
plt.title(f'Math Score vs Writing Score with Trendline - StudentID: {student_id} - {current_datetime}')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Select columns for Correlation Matrix
selected_columns = ['math score', 'reading score', 'writing score']

#Calculate Correlation Matrix
correlation_matrix = df[selected_columns].corr()

#Create Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title(f'Heatmap of Correlations - StudentID: {student_id} - {current_datetime}')
plt.show()
```
